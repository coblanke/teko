```python
import math
```

```python
def Bogenmass(alpha):
    phi = 2*math.pi/360 * alpha
    return phi
```

```python
def Gradmass(phi):
    alpha = 360/(2*math.pi) * phi
    return alpha
```

```python
def Polarkoordinate(x,y):
    r = math.sqrt(x**2+y**2)
    if x==0 and y==0:
        phi = 0.0
    elif x==0 and y>0:
        phi = math.pi/2
    elif x==0 and y<0:
        phi = 3*math.pi/2
    else:
        phi = math.atan(y/x)    # Fallunterscheidung!
    return (r, phi)
```

```python
def kartKoordinate(r, phi):
    x = r * math.cos(phi)
    y = r * math.sin(phi)
    return (x,y)
```

```python
def Drehung(x, y, alpha):
    (r,phi) = Polarkoordinate(x,y)
    (xneu, yneu) = kartKoordinate(r, phi+Bogenmass(alpha))
    return (xneu, yneu)
```

```python
Drehung(0,1,60)
```

```python
Punkte =[(1, 1), (0,1)]

import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = 'iframe'

fig = go.Figure()
for n in range(0, 361):
    testList=[]
    for (x,y) in Punkte:
        testList.append(Drehung(x, y, n))
    x_coords, y_coords = zip(*testList)
    fig.add_trace(
        go.Scatter(
            visible=False,
            x=x_coords,
            y=y_coords,
            name=f"N = {n}",
            mode="lines"
        )
    )

start_index = 0
fig.data[start_index].visible = True

# 5. Die Slider-Schritte (Steps) für JavaScript definieren
steps = []
for i in range(len(fig.data)):
    step = dict(
        method="update",
        args=[{"visible": [False] * len(fig.data)},  # Alle Kurven verstecken
              {"title": f"Live-Berechnung: x * ({i} * 2)"}], # Titel anpassen
        label=str(i)  # Beschriftung auf dem Slider
    )
    step["args"][0]["visible"][i] = True  # Nur die aktuelle Kurve einblenden
    steps.append(step)

# 6. Den Slider an das Layout der Grafik binden
sliders = [dict(
    active=start_index,
    currentvalue={"prefix": "Wert für N: "},
    pad={"t": 50},
    steps=steps
)]

fig.update_layout(
    sliders=sliders,
    title=f"Live-Berechnung: x * ({start_index} * 2)",
    xaxis_range=[-4,4], xaxis_title="X-Achse",
    yaxis_range=[-4,4], yaxis_title="Y-Achse",
    autosize=False,
    width=800,
    height=800,
)

fig.update_yaxes(
    scaleanchor="x",
    scaleratio=1,
  )

fig.show()
```
